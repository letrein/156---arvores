```python
import pandas as pd
from IPython.display import display

caminho = input("Digite o caminho do arquivo CSV: ")

df = pd.read_csv(caminho, sep=None, engine="python")

display(df)

try:
    df.to_clipboard(index=False, excel=True)
    print("Dados copiados. Cole-os no arquivo de destino com Ctrl+V.")
except Exception as erro:
    print(f"Não foi possível copiar para a área de transferência: {erro}")
    print("Você ainda pode salvar uma cópia em CSV:")
    df.to_csv("dados_copiados.csv", index=False)
```
